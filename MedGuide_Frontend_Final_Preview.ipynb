```python

%cd /content/Medical-RAG-Hallucination-Detection/frontend

from pathlib import Path

api_file = Path("src/services/api.ts")
lines = api_file.read_text().splitlines()

default_url = "http" + "://" + "127.0.0.1:8000"
replacement = (
    'const API_BASE_URL = import.meta.env.VITE_API_BASE_URL ?? "'
    + default_url
    + '";'
)

matches = [i for i, line in enumerate(lines) if line.startswith("const API_BASE_URL =")]

if len(matches) == 1:
    lines[matches[0]] = replacement
    api_file.write_text("\n".join(lines) + "\n")
    print("Updated the API URL line.")
else:
    print(f"Expected one API_BASE_URL declaration; found {len(matches)}. No change made.")

print("\n=== EXACT LINE NOW ON DISK ===")
for line in api_file.read_text().splitlines():
    if line.startswith("const API_BASE_URL ="):
        print(repr(line))
        print("Contains Markdown link syntax:", "](" in line)
```
